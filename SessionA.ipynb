In this session we will learn how to summarise data using measures of central tendency (mean, median, mode) and measures of spread (range, IQR, standard deviation). This helps reduce thousands of data points into a small set of meaningful numbers.

We will work with the penguins dataset, a fun dataset containing measurements of Antarctic penguins.

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

## Summary Statistics

Now we will calculate:

- **Mean**

We start with the variable `bill_length_mm`.

```python
penguins['bill_length_mm'].mean(), penguins['bill_length_mm'].median()
```

## Measures of Spread

We now calculate:

- **Range**  
- **Interquartile range (IQR)**  
- **Standard deviation**

```python
flipper = penguins['flipper_length_mm'].dropna()

# Range
data_range = flipper.max() - flipper.min()

# IQR
IQR = flipper.quantile(0.75) - flipper.quantile(0.25)

# Standard deviation
std_dev = flipper.std()

data_range, IQR, std_dev
```

## Creating a Summary Table

Let's generate a table showing the mean and standard deviation for  
`bill_length_mm` and `flipper_length_mm`.

```python
summary_table = penguins[['bill_length_mm', 'flipper_length_mm']].agg(['mean', 'std'])
summary_table
```
