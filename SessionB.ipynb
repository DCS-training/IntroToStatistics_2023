```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")
```

## Loading the Dataset

We load the penguins dataset from seaborn.

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

## Histograms — Identifying Distribution Shape

A histogram is the simplest tool for visualising the distribution of a variable.

Below we create a histogram of `bill_length_mm`.

```python
plt.figure(figsize=(7,4))
sns.histplot(data=penguins, x="bill_length_mm", bins=20, color="steelblue")
plt.show()
```

## 2. Density Plots — Smooth Distribution Shapes

A density plot gives a smooth approximation of a histogram.

```python
plt.figure(figsize=(7,4))
sns.kdeplot(data=penguins, x="bill_length_mm", fill=True, alpha=0.4, color="orange")
plt.show()
```

## 3. Comparing Groups Using Faceted Histograms

We often want to compare the distribution of a variable across groups.

Here you will create separate histograms of bill length for each penguin species.

```python
g = sns.FacetGrid(penguins, col="species", height=4, sharex=True, sharey=True)
g.map_dataframe(sns.histplot, x="bill_length_mm", bins=20, color="steelblue")
plt.show()
```

## 4. Scatterplots — Visualising Relationships

Scatterplots help reveal patterns and relationships between variables.

## 5. Normality Checking Using Q–Q Plots

Python does not include Q–Q plots in seaborn, so we use scipy.stats.

```python
import scipy.stats as stats
```

A Q–Q plot helps determine whether a variable follows a normal distribution.
If the points lie close to the diagonal line, the distribution is approximately normal.

```python
plt.figure(figsize=(6,6))
stats.probplot(penguins["bill_length_mm"].dropna(), dist="norm", plot=plt)
plt.title("Q-Q Plot: Bill Length")
plt.show()
```
